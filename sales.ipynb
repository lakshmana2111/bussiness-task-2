```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("/content/superstore_raw.csv")
```

```python
df.head()
```

```python
category_sales = df.groupby('Category')['Sales'].sum().reset_index()

plt.figure(figsize=(12, 6))
sns.barplot(x='Category', y='Sales', data=category_sales, hue='Category', palette='viridis', legend=False)
plt.title('Total Sales by Product Category')
plt.xlabel('Product Category')
plt.ylabel('Total Sales')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.boxplot(y='Profit', data=df)
plt.title('Profit Distribution')
plt.ylabel('Profit')
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(x='Discount', y='Profit', data=df, alpha=0.6)
plt.title('Discount vs. Profit')
plt.xlabel('Discount')
plt.ylabel('Profit')
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

```python
numerical_df = df.select_dtypes(include=['number'])
correlation_matrix = numerical_df.corr()

plt.figure(figsize=(10, 8))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Matrix of Numerical Columns')
plt.show()
```

```python
category_profit = df.groupby('Category')['Profit'].sum().reset_index()
highest_profit_category = category_profit.loc[category_profit['Profit'].idxmax()]

print("Category with the highest total profit:")
print(highest_profit_category)
```
